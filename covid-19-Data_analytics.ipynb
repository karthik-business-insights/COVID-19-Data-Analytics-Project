```python
import pandas as pd
day_wise=pd.read_csv(r"C:\Users\rangu\Downloads\Documents\covid-19\day_wise.csv")
display(day_wise)
country_wise=pd.read_csv(r"C:\Users\rangu\Downloads\Documents\covid-19\country_wise_latest.csv")
print("\n\n\n")
print("this is country wise data")
display(country_wise)
print(country_wise["Country/Region"].unique())
print("\n\n")
print("worldometer")
worldometer=pd.read_csv(r"C:\Users\rangu\Downloads\Documents\covid-19\worldometer_data.csv")
display(worldometer)

for c in country_wise["Country/Region"]:
    print(c)
    print("information about the couuntry wise data")
country_wise.info()
print("description about the country wise")
country_wise.describe()
print("is there any null values and counting")
country_wise.isnull().sum()
print("information about the worldometer data")
worldometer.info()
print("description about the worldometer data ")
worldometer.describe()
print("is there any null values and counting")
print(worldometer.isnull().sum())
worldometer["Continent"]=worldometer["Continent"].fillna("UNKNOWN")
print(worldometer["Continent"].isnull().sum())
print(worldometer.isnull().sum())
worldometer["Population"]=worldometer["Population"].fillna(worldometer["Population"].median())
print(worldometer["NewCases"].mean())
worldometer["NewCases"]=worldometer["NewCases"].fillna(worldometer["NewCases"].mean())
worldometer["TotalDeaths"]=worldometer["TotalDeaths"].fillna(worldometer["TotalDeaths"].mean())
print(worldometer["NewRecovered"].mean())
worldometer["NewRecovered"]=worldometer["NewRecovered"].fillna(worldometer["NewRecovered"].mean())
worldometer["ActiveCases"]=worldometer["ActiveCases"].fillna(worldometer["ActiveCases"].mean())
worldometer["Serious,Critical"]=worldometer["Serious,Critical"].fillna(worldometer["Serious,Critical"].mean())
worldometer["Tot Cases/1M pop"]=worldometer["Tot Cases/1M pop"].fillna(worldometer["Tot Cases/1M pop"].mean())
worldometer["Deaths/1M pop"]=worldometer["Deaths/1M pop"].fillna(worldometer["Deaths/1M pop"].mean())
worldometer["TotalTests"]=worldometer["TotalTests"].fillna(worldometer["TotalTests"].mean())
worldometer["Tests/1M pop"]=worldometer["Tests/1M pop"].fillna(worldometer["Tests/1M pop"].mean())
worldometer["WHO Region"]=worldometer["WHO Region"].dropna()
worldometer["NewDeaths"]=worldometer["NewDeaths"].fillna(worldometer["NewDeaths"].mean())
worldometer["TotalRecovered"]=worldometer["TotalRecovered"].fillna(worldometer["TotalRecovered"].mean())
print(worldometer.isnull().sum())
worldometer.dropna(subset=["WHO Region"], inplace=True)
print(worldometer["WHO Region"].isna().sum())
print(worldometer.isnull().sum())

print("\n\n")
print("data cleaning in country_wise ")
display(day_wise.head())
display(day_wise.info())
day_wise["Date"]=pd.to_datetime(day_wise["Date"])
print( day_wise.isnull().sum())
with pd.ExcelWriter("COVID19_Cleaned_Data.xlsx") as writer:
    day_wise.to_excel(writer, sheet_name="Day_Wise", index=False)
    country_wise.to_excel(writer, sheet_name="Country_Wise", index=False)
    worldometer.to_excel(writer, sheet_name="Worldometer", index=False)

print("Excel file saved successfully!")

import os
print(os.getcwd())



```
